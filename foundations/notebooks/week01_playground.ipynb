```python
# Trading twin: Form 4 transaction codes — the Box-7 checker's twin.
# Form 4 = the SEC filing a company insider (officer, director, 10% owner)
# files within two business days of trading the company's stock.
code = input("Form 4 transaction code: ").strip().upper()

if code == "P":
    meaning = "Open-market or private purchase"
elif code == "S":
    meaning = "Open-market or private sale"
elif code == "A":
    meaning = "Grant or award from the company"
elif code == "M":
    meaning = "Exercise or conversion of a deritave (e.g. stock options)"
elif code == "F":
    meaning = "Shares withheld to pay an exercise price or tax"
elif code == "G":
    meaning = "Gift"
else:
    meaning = "Not in this checker yet - add it"

# AFC's T1 trigger counts code P only: the insider spending their own money.
# A, M and F are compensation mechanics, not conviction.
# (P also covers private purchases; AFC v9.1 treats every P as its buy signal.)
is_open_market_purchase = code == "P"
print(f"{code}: {meaning} | counts for AFC T1: {is_open_market_purchase}")
```
