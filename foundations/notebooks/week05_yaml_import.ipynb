```python
import json

import yaml

from learning_journey.paths import PROJECT_ROOT

with open(PROJECT_ROOT / ".pre-commit-config.yaml", encoding="utf-8") as f:
    yaml_content = yaml.safe_load(f)

print("=" * 50)
print("Loaded YAML content in Python")
print("=" * 50)
print(yaml_content)

print("\n")
print("=" * 50)
print("Print out as YAML format")
print("=" * 50)
print(yaml.dump(yaml_content, indent=2))


print("\n")
print("=" * 50)
print("Print out as JSON format")
print("=" * 50)
# Optional flags:
# sort_keys=True sorts the keys alphabetically.
# ensure_ascii=False keeps accents readable (á, ñ).
# default=str prevents errors if YAML loaded dates, which JSON can't serialize.
print(json.dumps(yaml_content, indent=2, ensure_ascii=False, default=str))
```
